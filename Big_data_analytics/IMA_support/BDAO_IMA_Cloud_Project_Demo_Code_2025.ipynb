# **Option 1: Download data from Google Cloud Storage directly**

- Business data – gs://yelp-data-ima/yelp_academic_dataset_business_processed.json
- Check-in data – gs://yelp-data-ima/yelp_academic_dataset_checkin.json
- User data – gs://yelp-data-ima/yelp_academic_dataset_user.json
- Tips data – gs://yelp-data-ima/yelp_academic_dataset_tip.json
- Review data - gs://yelp-data-ima/yelp_academic_dataset_review.json

The following codes show you and example of downloading data from Google Cloud Storage using the provided uri. The uri follows the patter: gs://bucket_name/blob_name. You can replace the blob name (with data you would like to download) in the following codes to download respective dataset.

```python
# In this example, we download the business dataset from GCS

from google.cloud import storage
import pandas as pd
from io import StringIO

# Replace with your actual bucket and blob names
bucket_name = "yelp-data-ima"
blob_name = "yelp_academic_dataset_review.json"

# Initialize a Google Cloud Storage client
storage_client = storage.Client()

# Get the bucket and blob
bucket = storage_client.bucket(bucket_name)
blob = bucket.blob(blob_name)

# Download the blob's content as a string
data = blob.download_as_string()

# Decode the JSON data and read it into a pandas DataFrame
df = pd.read_json(StringIO(data.decode('utf-8')), orient='record', lines=True)

# Now you can work with the DataFrame 'df'
df.head()
```

# **Option 2: Load data into BigQuery**

Instead, you can load data into data warehouse for easy data query. The following codes help you load data from bucket to your BigQuery.

### Step 1: Create a new dataset in your BigQuery to store your project data

```python
# Import the necessary libraries
from google.cloud import storage
from google.cloud import bigquery
```

```python
# Create a dataset in GBQ to load tables
def create_dataset(project_id, dataset_id, location="US"):
    client = bigquery.Client(project=project_id)
    dataset_ref = bigquery.DatasetReference(project_id, dataset_id)
    dataset = bigquery.Dataset(dataset_ref)
    dataset.location = location

    created_dataset = client.create_dataset(dataset, timeout=30)  # Waits 30 seconds
    print("Dataset {} created.".format(created_dataset.dataset_id))

# Replace 'your-new-datatset-name' with your desired dataset name
create_dataset(project_id, 'your-new-datatset-name')
```

#### 01 Load business dataset into GBQ

```python
dataset_name = "your-dataset-name" # Replace 'your-dataset-name' with the dataset name you created
table_name = "your-table-name" # Replace 'your-table-name' with your desired table name
bucket_name = "yelp-data-ima" # don't make changes here
blob_name = "yelp_academic_dataset_business_processed.json" # don't make changes here

bigquery_client = bigquery.Client(project=project_id)

dataset_ref = bigquery_client.dataset(dataset_name)
table_ref = dataset_ref.table(table_name)

job_config = bigquery.LoadJobConfig()
job_config.autodetect = True
job_config.source_format = bigquery.SourceFormat.NEWLINE_DELIMITED_JSON

load_job = bigquery_client.load_table_from_uri(
    f"gs://{bucket_name}/{blob_name}",
    table_ref,
    job_config=job_config
)

load_job.result()

print(f"Data loaded into BigQuery: {project_id}.{dataset_name}.{table_name}")
```

### Joining tables through SQL query (example)

```python
# First make connection to the table in GBQ
from google.cloud import storage
from google.cloud import bigquery

dataset_name = "your-dataset-name" # Replace 'your-dataset-name' with the dataset name you created
table_1 = "your-table-name" # Replace 'your-table-name' with the name of the table you want to join
table_2 = "your-table-name" # Replace 'your-table-name' with the name of the table you want to join

bigquery_client = bigquery.Client(project=project_id)
```

You need to check the ERD (entity relationship diagram) before merging data. Business dataset, tips dataset, check-in dataset and review dataset can be joined by business_id. However, tips dataset and review dataset all have multiple entries for one business, which means you need to map the business information from business dataset to tips or review dataset if you want the merge. Review data and user data can be connected by user_id. You can map the user information to review dataset if you want.

```python
# Below is the example to merge business dataset and check-in dataset
query = f"""
    SELECT r.business_id, r.name, r.address, r.city, r.state, r.stars, r.review_count, r.is_open, r.categories,c.business_id, c.date
    FROM `{project_id}.{dataset_name}.{table_2}` AS c
    JOIN `{project_id}.{dataset_name}.{table_1}` AS r on r.business_id = c.business_id
    WHERE r.review_count>1000
"""

query_job = bigquery_client.query(query)
results = query_job.result()

join_df = query_job.to_dataframe()
join_df.head()
```
